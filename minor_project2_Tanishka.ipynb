```python
import pandas as pd
import numpy as np

# Load original dataset
df = pd.read_csv("rahul_transactions.csv")

print("=" * 60)
print("TRANSACTION PARSER")
print("=" * 60)

# Original row count
original_count = len(df)

# Count exact duplicate rows
duplicate_count = df.duplicated().sum()

# Remove exact duplicates
df = df.drop_duplicates().copy()

# Parse dates
df["date"] = pd.to_datetime(
    df["Date"],
    errors="coerce",
    dayfirst=True,
    format="mixed"
)

# Clean amount
df["amount"] = (
    df["Amount"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace("Rs.", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["amount"] = pd.to_numeric(
    df["amount"],
    errors="coerce"
)

# Standardise transaction type
df["type_clean"] = (
    df["Type"]
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({
        "dr": "debit",
        "debit": "debit",
        "cr": "credit",
        "credit": "credit"
    })
)

# Empty Mode → missing
df["Mode"] = df["Mode"].replace("", np.nan)

# Date/time features
df["hour"] = (
    df["Time"]
    .astype(str)
    .str[:2]
    .astype(int)
)

df["month"] = df["date"].dt.month

df["month_name"] = df["date"].dt.strftime("%b")

df["day_of_week"] = df["date"].dt.day_name()

df["week_number"] = df["date"].dt.isocalendar().week

print("Original transactions :", original_count)
print("Duplicates dropped    :", duplicate_count)
print("Clean transactions    :", len(df))
print("Unparseable dates     :", df["date"].isna().sum())
print("Unparseable amounts   :", df["amount"].isna().sum())

print("\nType counts:")
print(df["type_clean"].value_counts())
```

```python
vendor_keywords = {

    # Food Delivery
    "Swiggy": [
        "SWIGGY",
        "BUNDL"
    ],

    "Zomato": [
        "ZOMATO"
    ],

    # Quick Commerce
    "Swiggy Instamart": [
        "SWIGGY-INSTAMART",
        "BUNDL TECH-INSTAMART",
        "SWIGGY-INSTAMART",
        "INSTAMART"
    ],

    "Zepto": [
        "ZEPTO",
        "KIRANAKART"
    ],

    "Blinkit": [
        "BLINKIT",
        "GROFERS"
    ],

    # E-commerce
    "Amazon": [
        "AMAZON",
        "AMZN"
    ],

    "Flipkart": [
        "FLIPKART",
        "FKART"
    ],

    "Myntra": [
        "MYNTRA"
    ],

    "Nykaa": [
        "NYKAA",
        "FSN E-COMMERCE"
    ],

    # Groceries
    "BigBasket": [
        "BIGBASKET",
        "INNOVATIVE RETAIL"
    ],

    "DMart": [
        "DMART",
        "AVENUE SUPERMARTS"
    ],

    # Transport
    "Ola": [
        "OLA ELECTRIC",
        "OLA-PRIME",
        "OLACABS",
        "OLA CABS",
        "ANI TECHNOLOGIES"
    ],

    "Uber": [
        "UBER"
    ],

    "Rapido": [
        "RAPIDO",
        "ROPPEN TRANSPORTATION"
    ],

    "BMTC": [
        "BMTC",
        "TUMMOC"
    ],

    # Cafe
    "Starbucks": [
        "STARBUCKS",
        "TATA STARBUCKS"
    ],

    "Cafe Coffee Day": [
        "COFFEE DAY",
        "CCD"
    ],

    "Third Wave Coffee": [
        "THIRD WAVE",
        "THIRDWAVE",
        "TWC INDIA"
    ],

    # Restaurants
   "Restaurant": [
    "TRUFFLES",
    "EMPIRE RESTAURANT",
    "MEGHANA FOODS",
    "DINEOUT",
    "BANGALORE RESTAURANT",
    "UPI-RESTAURANT"
],

    # Subscriptions
    "Netflix": [
        "NETFLIX"
    ],

    "Spotify": [
        "SPOTIFY"
    ],

    "Disney+ Hotstar": [
        "DISNEY HOTSTAR",
        "HOTSTAR",
        "STAR INDIA PVT LTD"
    ],

    # Entertainment
    "BookMyShow": [
        "BOOKMYSHOW",
        "BMS MOVIE TICKETS",
        "BIGTREE ENTERTAINMENT"
    ],

   "Jio": [
    "JIOFIBER",
    "RELIANCE JIO",
    "JIORECHARGE"
],
    "Airtel": [
        "AIRTEL",
        "BHARTI AIRTEL"
    ],

    "Vi": [
        "VODAFONE IDEA",
        "VI POSTPAID",
        "UPI-VI-RECHARGE"
    ],

    "BESCOM": [
        "BESCOM",
        "BANGALORE ELEC SUPPLY"
    ],

    "BWSSB": [
        "BWSSB"
    ],

    # Investments
    "Zerodha": [
        "ZERODHA"
    ],

    "Groww": [
        "GROWW",
        "NEXTBILLION"
    ],

    # Fuel
    "BPCL": [
        "BPCL"
    ],

    "HP Petrol": [
        "HP PETROL"
    ],

    "Indian Oil": [
        "INDIAN OIL",
        "IOC"
    ],

    # Special transaction types
    "Cash Withdrawal": [
        "ATM-WDL"
    ],

    "P2P Transfer": [
        "UPI-AMAN-",
        "UPI-ANKIT-",
        "UPI-KARAN-",
        "UPI-NEHA-",
        "UPI-PRIYA-",
        "UPI-SNEHA-",
        "UPI-VIKAS-"
    ],

    "Rent": [
        "IMPS-RENT-LANDLORD"
    ],

    "Salary": [
        "TECHCRUSH LABS",
        "SALARY"
    ]
}
def extract_vendor(description):

    description = str(description).upper()

    for vendor, keywords in vendor_keywords.items():

        for keyword in keywords:

            if keyword in description:
                return vendor

    return "Uncategorised"


df["vendor_clean"] = df["Description"].apply(extract_vendor)

print("=" * 60)
print("VENDOR EXTRACTION")
print("=" * 60)

print("Uncategorised:", (df["vendor_clean"] == "Uncategorised").sum())

print("\nVendor counts:")
print(df["vendor_clean"].value_counts())
```

```python
category_map = {

    "Swiggy": "Food Delivery",
    "Zomato": "Food Delivery",

    "Swiggy Instamart": "Quick Commerce",
    "Zepto": "Quick Commerce",
    "Blinkit": "Quick Commerce",

    "Amazon": "E-commerce",
    "Flipkart": "E-commerce",
    "Myntra": "E-commerce",
    "Nykaa": "E-commerce",

    "Ola": "Transport",
    "Uber": "Transport",
    "Rapido": "Transport",
    "BMTC": "Transport",

    "Starbucks": "Cafe",
    "Cafe Coffee Day": "Cafe",
    "Third Wave Coffee": "Cafe",

    "Restaurant": "Restaurants",

    "Netflix": "Subscriptions",
    "Spotify": "Subscriptions",
    "Disney+ Hotstar": "Subscriptions",

    "Jio": "Utilities",
    "Airtel": "Utilities",
    "Vi": "Utilities",
    "BESCOM": "Utilities",
    "BWSSB": "Utilities",

    "BigBasket": "Groceries",
    "DMart": "Groceries",

    "Zerodha": "Investments",
    "Groww": "Investments",

    "BPCL": "Fuel",
    "HP Petrol": "Fuel",
    "Indian Oil": "Fuel",

    "BookMyShow": "Entertainment",

    "P2P Transfer": "Personal Transfer",
    "Cash Withdrawal": "Cash Withdrawal"
}

df["category"] = df["vendor_clean"].map(category_map)
def is_amazon_subscription(description):

    description = str(description).upper()

    return (
        "AMAZON PRIME" in description
        or "AMZN PRIME" in description
        or "AMAZON-PRIME" in description
        or "AMAZON PRIME VIDEO" in description
    )


amazon_subscription = df["Description"].apply(
    is_amazon_subscription
)

df.loc[
    amazon_subscription,
    "category"
] = "Subscriptions"
print("=" * 60)
print("CATEGORY AUDIT")
print("=" * 60)

print("\nCategory counts:")
print(df["category"].value_counts())

print("\nUncategorised transactions:")
print(df["category"].isna().sum())

print("\nUncategorised vendors:")
print(
    df.loc[
        df["category"].isna(),
        "vendor_clean"
    ].value_counts()
)
```

```python
print("=" * 60)
print("SPENDING OVERVIEW")
print("=" * 60)

total_credits = df[
    df["Type"].astype(str).str.strip().str.upper() == "CR"
]["amount"].sum()

total_debits = df[
    df["Type"].astype(str).str.strip().str.upper() == "DR"
]["amount"].sum()

net_change = total_credits - total_debits
savings_rate = (net_change / total_credits) * 100

merchant_vendor_count = df[
    ~df["vendor_clean"].isin(
        ["P2P Transfer", "Cash Withdrawal", "Salary", "Rent", "Uncategorised"]
    )
]["vendor_clean"].nunique()

print(f"Total Credits : ₹{total_credits:,.0f}")
print(f"Total Debits  : ₹{total_debits:,.0f}")
print(f"Net Change    : ₹{net_change:,.0f}")
print(f"Savings Rate  : {savings_rate:.1f}%")
print(f"Transactions  : {len(df):,}")
print(f"Unique Vendors: {merchant_vendor_count}")
```

```python
print("=" * 60)
print("TIME-OF-DAY PATTERNS")
print("=" * 60)

# Extract hour from Time
df["hour"] = df["Time"].astype(str).str[:2].astype(int)

# Food Delivery by hour
food_hour = (
    df[
        (df["category"] == "Food Delivery") &
        (df["type_clean"] == "debit")
    ]
    .groupby("hour")
    .size()
)

print("\nFood Delivery transactions by hour:")
print(food_hour)

# Cafe by hour
cafe_hour = (
    df[
        (df["category"] == "Cafe") &
        (df["type_clean"] == "debit")
    ]
    .groupby("hour")
    .size()
)

print("\nCafe transactions by hour:")
print(cafe_hour)

# Food Delivery between 9 PM and 1 AM
late_night = food_hour.reindex([21, 22, 23, 0], fill_value=0).sum()
total_food = food_hour.sum()

late_night_percentage = (late_night / total_food) * 100

print("\nLate-night Food Delivery (21:00-01:00):")
print(f"Transactions : {late_night}")
print(f"Total Food Delivery : {total_food}")
print(f"Percentage : {late_night_percentage:.2f}%")
print("=" * 60)
print("ANOMALY DETECTION")
print("=" * 60)

analysis_df = df[
    (df["type_clean"] == "debit") &
    (df["category"].notna())
].copy()

category_mean = analysis_df.groupby("category")["amount"].transform("mean")
category_std = analysis_df.groupby("category")["amount"].transform("std")

analysis_df["z_score"] = (
    (analysis_df["amount"] - category_mean) / category_std
)

anomalies = analysis_df[
    analysis_df["z_score"] > 2
].copy()

anomalies = anomalies.sort_values(
    "z_score",
    ascending=False
)

print("\nTotal anomalies:", len(anomalies))

print("\nTop anomalies:")
print(
    anomalies[
        [
            "Date",
            "Description",
            "amount",
            "vendor_clean",
            "category",
            "z_score"
        ]
    ]
    .head(30)
    .to_string(index=False)
)
```

```python
print("=" * 60)
print("ARCHETYPE DETECTION")
print("=" * 60)

# Category transaction counts
category_counts = (
    df[df["type_clean"] == "debit"]
    .groupby("category")
    .size()
)

# Category spending
category_spend = (
    df[df["type_clean"] == "debit"]
    .groupby("category")["amount"]
    .sum()
)

# Food Delivery late-night percentage
food_total = (
    df[
        (df["type_clean"] == "debit") &
        (df["category"] == "Food Delivery")
    ]
)

food_late_night = food_total[
    food_total["hour"].isin([21, 22, 23, 0])
]

food_late_pct = (
    len(food_late_night) / len(food_total) * 100
    if len(food_total) > 0 else 0
)

archetypes = []

# Foodie
if category_counts.get("Food Delivery", 0) >= 280:
    archetypes.append("Foodie")

# Quick Commerce Junkie
if category_counts.get("Quick Commerce", 0) >= 100:
    archetypes.append("Quick Commerce Junkie")

# Shopaholic
if category_counts.get("E-commerce", 0) >= 100:
    archetypes.append("Shopaholic")

# Investor
if category_counts.get("Investments", 0) >= 10:
    archetypes.append("Investor")

# Late-Night Snacker
if food_late_pct > 50:
    archetypes.append("Late-Night Snacker")

# YOLO Spender
if category_spend.sum() > 800000:
    archetypes.append("YOLO Spender")

print("\nDetected Archetypes:")
for archetype in archetypes:
    print("-", archetype)

print(f"\nFood Delivery late-night percentage: {food_late_pct:.2f}%")
print(f"Total debit-category spend: ₹{category_spend.sum():,.0f}")
```

```python
print("=" * 70)
print("             SpendDNA REPORT - RAHUL SHARMA")
print("        6 months - 1,310 transactions - Jan to Jun 2024")
print("=" * 70)

# ============================================================
# EXECUTIVE SUMMARY
# ============================================================

total_credits = df[
    df["Type"].astype(str).str.strip().str.upper() == "CR"
]["amount"].sum()

total_debits = df[
    df["Type"].astype(str).str.strip().str.upper() == "DR"
]["amount"].sum()

net_change = total_credits - total_debits
savings_rate = (net_change / total_credits) * 100

merchant_vendor_count = df[
    ~df["vendor_clean"].isin(
        ["P2P Transfer", "Cash Withdrawal", "Salary", "Rent", "Uncategorised"]
    )
]["vendor_clean"].nunique()

print("\nEXECUTIVE SUMMARY")
print("-" * 70)
print(f"  Total credits     : ₹{total_credits:,.0f}")
print(f"  Total debits      : ₹{total_debits:,.0f}")
print(f"  Net change        : ₹{net_change:,.0f}   (overspending)")
print(f"  Savings rate      : {savings_rate:.1f}%")
print(f"  Transactions      : {len(df):,}")
print(f"  Unique vendors    : {merchant_vendor_count}")


# ============================================================
# TOP CATEGORIES
# ============================================================

top_categories = (
    df[
        (df["Type"].astype(str).str.strip().str.upper() == "DR") &
        (df["category"].notna()) &
        (~df["category"].isin(
            ["Personal Transfer", "Cash Withdrawal"]
        ))
    ]
    .groupby("category")["amount"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print("\nTOP CATEGORIES (% of debit total)")
print("-" * 70)

max_category_amount = top_categories.max()

for category, amount in top_categories.items():
    percentage = (amount / total_debits) * 100

    bar_length = int(
        (amount / max_category_amount) * 20
    )

    bar = "#" * bar_length

    print(
        f"  {category:<18} "
        f"{bar:<20} "
        f"{percentage:>5.1f}%   "
        f"₹{amount:>10,.0f}"
    )


# ============================================================
# TOP VENDORS
# ============================================================

top_vendors = (
    df[
        (df["Type"].astype(str).str.strip().str.upper() == "DR") &
        (~df["vendor_clean"].isin(
            ["P2P Transfer", "Cash Withdrawal",
             "Salary", "Rent", "Uncategorised"]
        ))
    ]
    .groupby("vendor_clean")["amount"]
    .agg(["sum", "count"])
    .sort_values("sum", ascending=False)
    .head(5)
)

print("\nTOP VENDORS")
print("-" * 70)

for vendor, row in top_vendors.iterrows():
    print(
        f"  {vendor:<18} "
        f"₹{row['sum']:>10,.0f}   "
        f"({int(row['count'])} transactions)"
    )


# ============================================================
# TIME-OF-DAY PATTERNS
# ============================================================

food_hours = (
    df[
        (df["category"] == "Food Delivery") &
        (df["type_clean"] == "debit")
    ]
    .groupby("hour")
    .size()
)

cafe_hours = (
    df[
        (df["category"] == "Cafe") &
        (df["type_clean"] == "debit")
    ]
    .groupby("hour")
    .size()
)

peak_food_hour = food_hours.idxmax()
peak_food_count = food_hours.max()

peak_cafe_hour = cafe_hours.idxmax()
peak_cafe_count = cafe_hours.max()

late_night_food = food_hours.reindex(
    [21, 22, 23, 0],
    fill_value=0
).sum()

total_food_orders = food_hours.sum()

late_night_percentage = (
    late_night_food / total_food_orders
) * 100

print("\nTIME-OF-DAY PATTERNS")
print("-" * 70)
print(
    f"  Food Delivery peak : "
    f"{peak_food_hour:02d}:00 "
    f"({peak_food_count} transactions)"
)
print(
    f"  Cafe peak          : "
    f"{peak_cafe_hour:02d}:00 "
    f"({peak_cafe_count} transactions)"
)
print(
    f"  Food Delivery 21:00-01:00 : "
    f"{late_night_percentage:.2f}% of orders"
)


# ============================================================
# MONTHLY TREND — FOOD DELIVERY
# ============================================================

food_monthly = (
    df[
        (df["Type"].astype(str).str.strip().str.upper() == "DR") &
        (df["category"] == "Food Delivery")
    ]
    .groupby("month")["amount"]
    .sum()
    .reindex([1, 2, 3, 4, 5, 6], fill_value=0)
)

month_names = {
    1: "Jan",
    2: "Feb",
    3: "Mar",
    4: "Apr",
    5: "May",
    6: "Jun"
}

print("\nMONTHLY TREND (Food Delivery)")
print("-" * 70)

max_food_month = food_monthly.max()

for month, amount in food_monthly.items():

    bar_length = int(
        (amount / max_food_month) * 20
    )

    bar = "#" * bar_length

    print(
        f"  {month_names[month]:<4} "
        f"₹{amount:>10,.0f}   "
        f"{bar}"
    )


# ============================================================
# MONTH-ON-MONTH TREND
# ============================================================

monthly_trend = (
    df[
        (df["type_clean"] == "debit") &
        (df["category"].notna())
    ]
    .pivot_table(
        values="amount",
        index="category",
        columns="month",
        aggfunc="sum",
        fill_value=0
    )
    .reindex(
        columns=[1, 2, 3, 4, 5, 6],
        fill_value=0
    )
)

monthly_trend.columns = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun"
]

monthly_change = monthly_trend.pct_change(
    axis=1,
    fill_method=None
) * 100

previous_month = monthly_trend.shift(axis=1)

monthly_change = monthly_change.where(
    previous_month != 0
)

changes = monthly_change.stack()

biggest_increase = changes.idxmax()
biggest_decrease = changes.idxmin()
print(
    f"\n  Biggest overall increase : "
    f"{biggest_increase[0]} ({biggest_increase[1]}) "
    f"{changes.loc[biggest_increase]:+.2f}%"
)

print(
    f"  Biggest overall decrease : "
    f"{biggest_decrease[0]} ({biggest_decrease[1]}) "
    f"{changes.loc[biggest_decrease]:+.2f}%"
)

# ============================================================
# TOP ANOMALIES
# ============================================================

print("\nTOP ANOMALIES")
print("-" * 70)

for _, row in anomalies.head(5).iterrows():
    date_display = row["date"].strftime("%d %b")

    print(
        f"  {date_display:<8} "
        f"{row['vendor_clean']:<18} "
        f"₹{row['amount']:>8,.0f} "
        f"(z={row['z_score']:.2f})"
    )


# ============================================================
# ARCHETYPES
# ============================================================

category_counts = (
    df[df["type_clean"] == "debit"]
    .groupby("category")
    .size()
)

category_spend = (
    df[df["type_clean"] == "debit"]
    .groupby("category")["amount"]
    .sum()
)

archetypes = []

if category_counts.get("Food Delivery", 0) >= 280:
    archetypes.append("THE FOODIE")

if category_counts.get("Quick Commerce", 0) >= 100:
    archetypes.append("THE QUICK COMMERCE JUNKIE")

if category_counts.get("E-commerce", 0) >= 100:
    archetypes.append("THE SHOPAHOLIC")

if category_counts.get("Investments", 0) >= 10:
    archetypes.append("THE INVESTOR")

if late_night_percentage > 50:
    archetypes.append("THE LATE-NIGHT SNACKER")

if category_spend.sum() > 800000:
    archetypes.append("THE YOLO SPENDER")


# ============================================================
# ARCHETYPE DISPLAY
# ============================================================

print("\nRAHUL'S SPENDING ARCHETYPES")
print("-" * 70)

for archetype in archetypes:
    print(f"  -> {archetype}")


# ============================================================
# KEY INSIGHTS
# ============================================================

ecommerce_spend = top_categories.get(
    "E-commerce", 0
)

food_spend = top_categories.get(
    "Food Delivery", 0
)

quick_spend = (
    df[
        (df["Type"].astype(str).str.strip().str.upper() == "DR") &
        (df["category"] == "Quick Commerce")
    ]["amount"].sum()
)

investments_spend = top_categories.get(
    "Investments", 0
)

combined_discretionary = (
    ecommerce_spend +
    food_spend +
    quick_spend
)

combined_percentage = (
    combined_discretionary /
    total_debits
) * 100

print("\nKEY INSIGHTS")
print("-" * 70)

print(
    f"  1. E-commerce is the largest DR spending "
    f"category at ₹{ecommerce_spend:,.0f} "
    f"({(ecommerce_spend / total_debits) * 100:.1f}% of debits)."
)

print(
    f"  2. Food Delivery recorded {int(total_food_orders)} "
    f"transactions, with the peak occurring at "
    f"{peak_food_hour:02d}:00."
)

print(
    f"  3. E-commerce + Food Delivery + Quick Commerce "
    f"account for ₹{combined_discretionary:,.0f}, "
    f"or {combined_percentage:.1f}% of DR spending."
)

print(
    f"  4. Investments account for "
    f"₹{investments_spend:,.0f} "
    f"({(investments_spend / total_debits) * 100:.1f}% "
    f"of DR spending)."
)

print(
    f"  5. {len(anomalies)} transactions were flagged "
    f"as anomalies using z > 2."
)


# ============================================================
# FINAL
# ============================================================

print("\n" + "=" * 70)
print("                         END REPORT")
print("=" * 70)
```
